```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Mini Data Analysis Project

Goal: To analyze student performance using the same DataFrame i've beem practicing with

```python
students = pd.DataFrame({
    "Gender": ["Female", "Male", "Male", "Female", "Female", "Male", "Female", "Male"],
        "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
        "Exam_Score": [35, 50, 65, 80, 95, 96, 97, 98],
        "Hours_Slept": [6, 7, 5, 8, 6, 9, 7, 8]
})

sns.scatterplot(
    data=students,
    x="Study_Hours",
    y="Exam_Score",
    hue="Gender"
)

plt.title("Study Hours vs Exam Score")
plt.show()
```

```python
sns.barplot(
    data=students,
    x="Gender",
    y="Exam_Score",
    errorbar=None
)

plt.title("Average Exam Score by Gender")
plt.show()
```

```python
sns.scatterplot(
    data=students,
    x="Hours_Slept",
    y="Exam_Score",
    hue="Gender"
)

plt.title("Hours Slept vs Exam Score")
plt.show()
```

# Dashboard

```python
fig, ax = plt.subplots(1, 2, figsize=(22, 11))

ax[0].scatter(
    students["Study_Hours"],
    students["Exam_Score"],
)

ax[0].set_xlabel("Score")
ax[0].set_ylabel("Gender")
ax[0].set_title("Score distribution across different gender")

sns.scatterplot(
    data=students,
    x="Study_Hours",
    y="Exam_Score",
    hue="Gender",
    ax=ax[1]
)

ax[1].set_xlabel("Score")
ax[1].set_ylabel("Hours Slept")
ax[1].set_title("Study Hours vs Exam Score")

plt.tight_layout()
plt.show()
```
