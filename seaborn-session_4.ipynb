```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Pair plot

```python
students = pd.DataFrame({
    "Gender": ["Female", "Male", "Male", "Female", "Female", "Male", "Female", "Male"],
        "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
        "Exam_Score": [35, 50, 65, 80, 95, 96, 97, 98],
        "Hours_Slept": [6, 7, 5, 8, 6, 9, 7, 8]
})

sns.pairplot(
    data=students,
    vars=["Study_Hours", "Exam_Score", "Hours_Slept"],
    hue="Gender"
)

plt.title("hello")
plt.show()
```

# Joint Plot


```python
sns.jointplot(
    data=students,
    x="Study_Hours",
    y="Exam_Score",
    kind="scatter"
)

plt.show()
```

# Replot

```python
sns.relplot(
    data=students,
    x="Study_Hours",
    y="Exam_Score",
    hue="Gender",
    col="Gender",
    kind="scatter"
)

plt.show()
```
