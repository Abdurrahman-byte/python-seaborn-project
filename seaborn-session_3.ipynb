```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Histplot

A histogram groups numerical values into intervals called bins and shows how many observations falls into each intervals.

```python
students = pd.DataFrame({
    "Gender": ["Female", "Male", "Male", "Female", "Female", "Male", "Female", "Male"],
        "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
        "Exam_Score": [35, 50, 65, 80, 95, 96, 97, 98],
        "Hours_Slept": [6, 7, 5, 8, 6, 9, 7, 8]
})

sns.histplot(
    data=students,
    x="Exam_Score",
    bins=5,
    hue="Gender"
)

plt.title("Distribution of Exam Scores")
plt.show()
```

# Box Plot

```python
sns.boxplot(
    data=students,
    x="Gender",
    y="Exam_Score",
)

plt.title("Exam Score Spread by Gender")
plt.show()
```

# Heat Maps

```python
correlation = students[
    ["Study_Hours", "Exam_Score", "Hours_Slept"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin="-1",
    vmax=1
)

plt.title("Student Data Correlation")
plt.show()
```
