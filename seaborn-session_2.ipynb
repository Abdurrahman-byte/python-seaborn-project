Line plot

```python
students = pd.DataFrame({
    "Gender": ["Female", "Male", "Male", "Female", "Female", "Male", "Female", "Male"],
        "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
        "Exam_Score": [35, 50, 65, 80, 95, 96, 97, 98],
        "Hours_Slept": [6, 7, 5, 8, 6, 9, 7, 8]
})

sns.lineplot(
    data=students,
    x="Study_Hours",
    y="Exam_Score"
)

plt.title("Study Hours vs Exam score")
plt.show()
```

Bar Plot

```python
sns.barplot(
    data=students,
    x="Gender",
    y="Exam_Score"
)



plt.title("Comparing averages")
plt.show()
```

Count Plot

```python
sns.countplot(
    data=students,
    x="Gender",
)

plt.title("Counting categories")
plt.show()
```
