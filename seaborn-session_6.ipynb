```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Facet Grid

```python
students = pd.DataFrame({
    "Gender": ["Female", "Male", "Male", "Female", "Female", "Male", "Female", "Male"],
        "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
        "Exam_Score": [35, 50, 65, 80, 95, 96, 97, 98],
        "Hours_Slept": [6, 7, 5, 8, 6, 9, 7, 8]
})

g = sns.FacetGrid(
    data=students,
    col="Gender"
)

g.map_dataframe(
    sns.scatterplot,
    x="Study_Hours",
    y="Exam_Score"
)

plt.show()
```

# Error bars

```python
sns.barplot(
    data=students,
    x="Gender",
    y="Exam_Score",
    errorbar=None
)

plt.show()
```

# Consistent Chart Styling

```python
sns.set_theme(
    style="whitegrid",
    palette="deep"
)
```
