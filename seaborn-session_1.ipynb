# Scatterplot

investigating the relationship between study hours and exam score

```python
data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Exam_Score": [35, 42, 50, 55, 65]
})

sns.scatterplot(
    data=data,
    x="Study_Hours",
    y="Exam_Score"
)

plt.title("Study Hours vs Exams Score")
plt.show()
```

```python
students = pd.DataFrame({
    "Gender": ["Female", "Male", "Male", "Female", "Female", "Male", "Female", "Male"],
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Exam_Score": [35, 50, 65, 80, 95, 96, 97, 98],
    "Hours_Slept": [6, 7, 5, 8, 6, 9, 7, 8]
})

sns.scatterplot(
    data=students,
    hue="Gender",
    x="Study_Hours",
    y="Exam_Score"
)

plt.title("Study Hours vs Exam Score")
plt.show()
```

# Style & Size

The style parameter distinguish categries by changing the marker shape instead of relying only on color

```python
sns.scatterplot(
    data=students,
    x="Study_Hours",
    y="Exam_Score",
    hue="Gender",
    size="Hours_Slept",
    sizes=(50, 250)
)
plt.title("Study Hours vs Exam Score")
plt.show()
```

```python
sns.scatterplot(
    data=students,
    x="Study_Hours",
    y="Exam_Score",
    hue="Gender",
    style="Gender",
    size="Hours_Slept",
    sizes=(50,250)
)

plt.title("Study Hours vs Exam Score")
plt.show()
```
